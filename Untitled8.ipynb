## • Handling Missing Values :

In this step, we have multiple data types.

* Some values ​​were replaced with "none" because they are extraneous featuers.

 * Numerical values ​​were filled in using the median

 * categorical values ​​using the mode.

 * Finally, I ensured that all missing values ​​were handled correctly.

```python
none_cols = ["PoolQC", "MiscFeature", "Alley", "Fence", "FireplaceQu"]

for col in none_cols:
    if col in train.columns:
        train[col] = train[col].fillna("None")


num_cols = train.select_dtypes(include=["int64", "float64"]).columns

for col in num_cols:
    train[col] = train[col].fillna(train[col].median())


cat_cols = train.select_dtypes(include=["object"]).columns

for col in cat_cols:
    train[col] = train[col].fillna(train[col].mode()[0])


train.isnull().sum().head(10)
```

## 4- Feature Engineering

In this step, I created a new feature called TotalSF by combining the basement, first floor, and second floor areas.
This feature represents the total living space of the house, which can help improve the model's understanding of house size.

```python
train["TotalSF"] = train["TotalBsmtSF"] + train["1stFlrSF"] + train["2ndFlrSF"]
test["TotalSF"] = test["TotalBsmtSF"] + test["1stFlrSF"] + test["2ndFlrSF"]
```

```python
qual_map = { "Ex": 5, "Gd": 4, "TA": 3, "Fa": 2,  "Po": 1 }

label_cols = ["ExterQual", "KitchenQual"]

for col in label_cols:
    train[col] = train[col].astype(str).str.strip().map(qual_map)
    test[col] = test[col].astype(str).str.strip().map(qual_map)
```

## •Encoding categorical variables .
I first selected categorical columns, then applied pd.get_dummies() to both train and test datasets. After that, I aligned both datasets to ensure they have the same columns and filled missing values with zeros. Finally, I verified the shapes of the datasets.

```python
nominal_cols = train.select_dtypes(include="object").columns

train = pd.get_dummies(train, columns=nominal_cols)
test = pd.get_dummies(test, columns=nominal_cols)
```

```python
train, test = train.align(test, join="left", axis=1, fill_value=0)
```
